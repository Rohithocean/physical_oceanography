## **Description**

This notebook provides a step-by-step guide to setting up, compiling, and running the **Regional Ocean Modeling System (ROMS)** — a widely used, free-surface, terrain-following ocean general circulation model.  
ROMS is extensively used to simulate and understand oceanic processes such as coastal dynamics, upwelling systems, thermohaline circulation, and regional climate variability.

---

The notebook is structured to cover the complete modeling pipeline:

4. **Model Compilation**
